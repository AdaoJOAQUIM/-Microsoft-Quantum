```python
# 1. Installation
!pip install -q transformers

# 2. Importation et chargement du modèle
from transformers import AutoTokenizer, AutoModelForCausalLM

model_id = "HuggingFaceTB/SmolLM2-360M-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

# 3. Préparer la question
messages = [
    {"role": "user", "content": "Bonjour, qui es-tu ?"}
]

inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    return_tensors="pt"
)

# 4. Générer la réponse
# Utilizar **inputs para passar os tensores corretamente
outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    do_sample=True,
    temperature=0.7
)

# 5. Afficher
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# 1. Installations
!pip install -q transformers gradio

# 2. Chargement du modèle
from transformers import AutoTokenizer, AutoModelForCausalLM
import gradio as gr

model_id = "HuggingFaceTB/SmolLM2-360M-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

# 3. Fonction de réponse
def repondre(message, history):
    messages = [{"role": "user", "content": message}]
    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        return_tensors="pt"
    )

    # Ajuste: usamos **inputs e acessamos o input_ids corretamente
    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=True,
        temperature=0.7
    )

    # Decodificar apenas a parte nova (a resposta do assistente)
    input_length = inputs['input_ids'].shape[1]
    reponse = tokenizer.decode(outputs[0][input_length:], skip_special_tokens=True)
    return reponse

# 4. Lancer l'interface
gr.ChatInterface(
    fn=repondre,
    title="SmolLM2-360M-Instruct",
    description="Pose une question au modèle"
).launch(share=True)
```
