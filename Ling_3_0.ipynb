# Ling 3.0 CPU Setup
Ce notebook installe et configure un modèle Ling (via Llama-CPP) pour fonctionner uniquement sur CPU.

```python
!pip install llama-cpp-python huggingface_hub
```

```python
from huggingface_hub import hf_hub_download
from llama_cpp import Llama

try:
    # Téléchargement de la version Instruct spécifique
    model_path = hf_hub_download(
        repo_id="bartowski/Llama-3.2-1B-Instruct-GGUF",
        filename="Llama-3.2-1B-Instruct-Q4_K_M.gguf"
    )

    # Chargement du modèle Instruct sur CPU
    llm = Llama(
        model_path=model_path,
        n_ctx=2048,
        n_threads=4,
        verbose=False
    )
    print("Version Instruct de Ling chargée avec succès.")
except Exception as e:
    print(f"Erreur : {e}")
```

```python
def chat_with_ling(prompt):
    if 'llm' not in globals():
        return "Erreur : Le modèle n'est pas chargé. Veuillez exécuter la cellule d00fe4c1."

    output = llm(
        f"Q: {prompt} A:",
        max_tokens=128,
        stop=["Q:", "\n"],
        echo=False
    )
    return output['choices'][0]['text']

# Test de conversation
try:
    user_input = "Bonjour, comment vas-tu ?"
    response = chat_with_ling(user_input)
    print(f"Ling: {response}")
except Exception as e:
    print(f"Erreur : {e}")
```

```python
!pip install gradio

import gradio as gr

def ling_chat_interface(message, history):
    # On réutilise la fonction de chat définie précédemment
    return chat_with_ling(message)

demo = gr.ChatInterface(
    fn=ling_chat_interface,
    title="Ling 3.0 CPU Chat",
    description="Interface de conversation pour le modèle Ling optimisé pour CPU."
)

demo.launch(share=True)
```
